```python
# Get Images to List
def get_image_paths(image_folder_path, extensions=('.jpg', '.jpeg', '.png', '.gif', '.bmp')):
    """Return a list of full file paths to all images in the specified folder."""
    all_files = os.listdir(image_folder_path)  # Get all file names in the folder
    image_files = [
        file_name for file_name in all_files
        if file_name.lower().endswith(extensions)  # Check if the file has a valid image extension
    ]
    full_paths = [os.path.join(image_folder_path, file_name) for file_name in image_files]
    return full_paths

# Example usage:
image_folder_path =  'AVA images/' # Replace with the path to your folder
image_paths = get_image_paths(image_folder_path)
image_paths
```

### Accessing the logprobs?

```python
import numpy as np
from IPython.display import display, HTML

def get_completion(
    messages: list[dict[str, str]],
    model: str = "gpt-4",
    max_tokens=500,
    temperature=0,
    stop=None,
    seed=123,
    tools=None,
    logprobs=None,  # whether to return log probabilities of the output tokens or not. If true, returns the log probabilities of each output token returned in the content of message..
    top_logprobs=None,
) -> str:
    params = {
        "model": model,
        "messages": messages,
        "max_tokens": max_tokens,
        "temperature": temperature,
        "stop": stop,
        "seed": seed,
        "logprobs": logprobs,
        "top_logprobs": top_logprobs,
    }
    if tools:
        params["tools"] = tools

    completion = client.chat.completions.create(**params)
    return completion
```

## Code for getting the image URLs given a dpchallenge Code

```python
from urllib.request import urlopen
from urllib.error import HTTPError
import re

def get_dpchallenge_image_url(image_id):
    """
    Given a dpchallenge image ID, retrieves the image page and extracts
    the image URL that ends with '<image_id>.jpg'.
    
    Args:
        image_id (int): The numeric ID of the image.
        
    Returns:
        str or None: The extracted image URL if found, otherwise None.
    """
    # Construct dp challenge URL
    page_url = f'https://www.dpchallenge.com/image.php?IMAGE_ID={image_id}'
    
    try:
        # Retrieve and decode HTML source code
        source = urlopen(page_url).read().decode('utf-8', errors='replace')
    except HTTPError as e:
        # Log the error and return None for this image_id
        print(f"Error {e.code} for image ID {image_id}: {e.reason}")
        return None
    
    # Build a regex pattern to match a URL ending with "<image_id>.jpg".
    pattern = r'(https?://[^"]+' + str(image_id) + r'\.jpg)'
    
    # Search pattern in source
    matches = re.findall(pattern, source)
    
    # Return the first match if available.
    return matches[0] if matches else None

# Test function 
get_dpchallenge_image_url(267110)
# get_dpchallenge_image_url(202619)
```
