### Accessing the logprobs?

```python
import numpy as np
from IPython.display import display, HTML

def get_completion(
    messages: list[dict[str, str]],
    model: str = "gpt-4",
    max_tokens=500,
    temperature=0,
    stop=None,
    seed=123,
    tools=None,
    logprobs=None,  # whether to return log probabilities of the output tokens or not. If true, returns the log probabilities of each output token returned in the content of message..
    top_logprobs=None,
) -> str:
    params = {
        "model": model,
        "messages": messages,
        "max_tokens": max_tokens,
        "temperature": temperature,
        "stop": stop,
        "seed": seed,
        "logprobs": logprobs,
        "top_logprobs": top_logprobs,
    }
    if tools:
        params["tools"] = tools

    completion = client.chat.completions.create(**params)
    return completion
```

```python
CLASSIFICATION_PROMPT = """You will be given a headline of a news article.
Please tell me whether the following tweet does contain offensive language. 
Please return me only the following: 
1) Return 'OS' if the tweet does contains Offensive language (if not return 'non-OS')
2) Return a a number from 1 to 4 representing your certainty rating for the decision (1 is low certainty, 4 is high certainty). 
Here is the tweet: {tweet}"""
```

```python
tweets = [
    "I live for two-bite brownies",
    "@###### see bitch this is why you have no friends",
    "I f*****g love this"
]
```

```python
API_RESPONSE.choices[0].message.content
```

```python
for tweet in tweets:
    print(f"\nTweet: {tweet}")
    API_RESPONSE = get_completion(
        [{"role": "user", "content": CLASSIFICATION_PROMPT.format(tweet=tweet)}],
        model="gpt-4o-mini",
        logprobs=True,
        top_logprobs=5,
        temperature=1
    )
    
    # Extract and print the response text
    response_text = API_RESPONSE.choices[0].message.content
    print("Response: " + response_text)
    
    top_two_logprobs = API_RESPONSE.choices[0].logprobs.content[0].top_logprobs
    html_content = ""
    for i, logprob in enumerate(top_two_logprobs, start=1):
        html_content += (
            f"<span style='color: cyan'>Output token {i}:</span> {logprob.token}, "
            f"<span style='color: darkorange'>logprobs:</span> {logprob.logprob}, "
            f"<span style='color: magenta'>linear probability:</span> {np.round(np.exp(logprob.logprob)*100,2)}%<br>"
        )
    display(HTML(html_content))
    print("\n")
```

```python
# logprob output as described here : https://cookbook.openai.com/examples/using_logprobs 
for image_url in images_urls:
    print(f"\nImage URL: {image_url}")
    API_RESPONSE = get_completion(
        [{"role": "user", "content": CLASSIFICATION_PROMPT.format(image_url=image_url)}],
        model="gpt-4o-mini",
        # model="gpt-4-turbo",
        logprobs=True,
        top_logprobs=10,
        temperature = 1 # default: 1 --> assoc. of logprobs & temp. not perfectly clear
    )
    
    # Extract and print the response text
    response_text = API_RESPONSE.choices[0].message.content
    print("Response: " + response_text)
    
    top_two_logprobs = API_RESPONSE.choices[0].logprobs.content[0].top_logprobs
    html_content = ""
    for i, logprob in enumerate(top_two_logprobs, start=1):
        html_content += (
            f"<span style='color: cyan'>Output token {i}:</span> {logprob.token}, "
            f"<span style='color: darkorange'>logprobs:</span> {logprob.logprob}, "
            f"<span style='color: magenta'>linear probability:</span> {np.round(np.exp(logprob.logprob)*100,2)}%<br>"
        )
    display(HTML(html_content))
    print("\n")
```

## Code for getting the image URLs given a dpchallenge Code

```python
from urllib.request import urlopen
from urllib.error import HTTPError
import re

def get_dpchallenge_image_url(image_id):
    """
    Given a dpchallenge image ID, retrieves the image page and extracts
    the image URL that ends with '<image_id>.jpg'.
    
    Args:
        image_id (int): The numeric ID of the image.
        
    Returns:
        str or None: The extracted image URL if found, otherwise None.
    """
    # Construct dp challenge URL
    page_url = f'https://www.dpchallenge.com/image.php?IMAGE_ID={image_id}'
    
    try:
        # Retrieve and decode HTML source code
        source = urlopen(page_url).read().decode('utf-8', errors='replace')
    except HTTPError as e:
        # Log the error and return None for this image_id
        print(f"Error {e.code} for image ID {image_id}: {e.reason}")
        return None
    
    # Build a regex pattern to match a URL ending with "<image_id>.jpg".
    pattern = r'(https?://[^"]+' + str(image_id) + r'\.jpg)'
    
    # Search pattern in source
    matches = re.findall(pattern, source)
    
    # Return the first match if available.
    return matches[0] if matches else None

# Test function 
get_dpchallenge_image_url(267110)
# get_dpchallenge_image_url(202619)
```

```python
# Loop through multiple IDs and generate a dataframe with output image URLs
pd.set_option('display.max_colwidth', None)

# sample_ids = [10013,267110,221721]
sample_ids = ava_ratings['image_num'].sample(n=500, random_state=1)

# Extract image URLs for each ID.
results = []
for image_id in sample_ids:
    url = get_dpchallenge_image_url(image_id)
    results.append({'ID': image_id, 'img_url': url})

# Create a DataFrame from the results.
df = pd.DataFrame(results)
df
```
